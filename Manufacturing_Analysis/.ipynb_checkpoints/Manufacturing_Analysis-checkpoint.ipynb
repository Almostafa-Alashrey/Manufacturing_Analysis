<div style="color: Green;  padding: 7px; border-radius: 5px; font-weight: bold; font-size: 25px;"> Data Cleaning </div>

---

```python
# Display Columns with Missing Values (> 0%)
missing_percentage = (manufacturing.isnull().mean().mul(100).round(2).sort_values(ascending=False))
missing_percentage = missing_percentage[missing_percentage > 0]
print(missing_percentage) 
```

#### `Observation`
-  The missing values represent  0% of the dataset , So the manufacturing dataset hasn't contain missing values  .

```python
# Identify duplicate records in the dataset.
manufacturing.duplicated().sum() 
```

#### `Observation`
- the manufacturing dataset hasn't contain duplicate values 

```python
# Count the number of unique values in each column.
manufacturing.nunique() 
```

```python
# Standardize column names by:Removing leading/trailing whitespace,Converting names to Title ,Replacing spaces and hyphens with underscores.
# snake_case 
manufacturing.columns = (manufacturing.columns.str.strip().str.replace(" " ,"_" , regex = False).str.replace("/" ,"_" , regex = False)
    .str.replace("-" ,"_" , regex = False).str.lower()) 
print(manufacturing.columns)
```

```python
# Replaced shift values with more understandable labels: Morning, Evening, and Night shifts.
manufacturing['shift'] = manufacturing['shift'].replace({'Shift_1':'Morning' ,'Shift_2': 'Evening' ,'Shift_3': 'Night'})
```

```python
# Dropped the production_efficiency column because it represents the same values as the quality column.
manufacturing.drop(columns='production_efficiency' , inplace = True )
```
